```python
import numpy as np
```

```python
#entenda uma classe como uma fabrica de objetos
class Vetor3d:

    #Propriedades genericas da classe
    #self é um place holder pro objeto que ainda nao foi criado (por isso o self)
    def __init__(self, i, j, k):
        self.i = i
        self.j = j
        self.k = k

    #Métodos da classe, ou seja, as ações que os objetos dessa classe podem realizar

    def escalar(self, v2: Vetor3d):
        if not isinstance(v2, Vetor3d):
            raise TypeError("O operando deve ser do tipo Vetor3d")
        return self.i * v2.i + self.j * v2.j + self.k * v2.k
    
    def vetorial(self, v2: Vetor3d):
        if not isinstance(v2, Vetor3d):
            raise TypeError("O operando deve ser do tipo Vetor3d")
        i = self.j * v2.k - self.k * v2.j
        j = self.k * v2.i - self.i * v2.k
        k = self.i * v2.j - self.j * v2.i
        return Vetor3d(i, j, k)

    def angulo(self, v2: 'Vetor3d', graus=False):
        if not isinstance(v2, Vetor3d):
            raise TypeError("O operando deve ser do tipo Vetor3d")
        if not isinstance(graus, bool):
            raise TypeError("o tipo deve ser do tipo boolean")
        
        produto_escalar = self.escalar(v2)
        cos_theta = produto_escalar / (self.norma() * v2.norma())
        theta = np.arccos(cos_theta)
        
        if graus:
            return np.degrees(theta)
        return theta

    def norma(self):
        return np.sqrt(self.i**2 + self.j**2 + self.k**2)

    def __add__(self, v2: Vetor3d):
        if not isinstance(v2, Vetor3d):
            raise TypeError("O operando deve ser do tipo Vetor3d")
        vetor_resultante = Vetor3d(i = self.i + v2.i,
                                   j = self.j + v2.j,
                                   k = self.k + v2.k)
        return vetor_resultante

    def __repr__(self):
        return f"{self.i}i + {self.j}j + {self.k}k"

    @staticmethod 
    #precisa ser declarado como staticmethod para que possa ser chamado sem a necessidade de criar um objeto da classe
    def geravetor3d(lim_inf: int, lim_sup: int): #metodo estatico, ou seja, um metodo que pode ser chamado sem a necessidade de criar um objeto da classe
    #poderiamos transformar esse metodo em uma função normal, mas como ele tem a ver com a classe, faz sentido que ele seja um metodo estatico da classe
        if not isinstance(lim_inf, int) or not isinstance(lim_sup, int):
            raise TypeError("Os limites devem ser do tipo inteiro")
        i = np.random.randint(lim_inf, lim_sup)
        j = np.random.randint(lim_inf, lim_sup)
        k = np.random.randint(lim_inf, lim_sup)
        return Vetor3d(i, j, k)
    
```
