O **Método dos Mínimos Quadrados** é uma técnica de aproximação numérica que encontra a curva polinomial que **melhor se ajusta** a um conjunto de pontos, minimizando a soma dos quadrados dos resíduos (diferenças entre os valores observados e os valores estimados pelo modelo).

---

$$P(x) = a_g x^g + a_{g-1} x^{g-1} + \cdots + a_1 x + a_0$$

tal que a soma seja mínima.

 $$S =  \frac{1}{2} \sum_{i=1}^{n} [y_i - P(x_i)]^2$$


```python
#condicoes de contorno
dados = np.array([[1,6], [2,12], [3,20]])
display(dados)

#graus de polinomio
g = 2
```

## Calculando as Somatórias (Lado Esquerdo do Sistema)

Para um polinômio de grau 2, o sistema normal do MMQ tem a forma:

$$
\begin{bmatrix}
\sum x^4 & \sum x^3 & \sum x^2 \\
\sum x^3 & \sum x^2 & \sum x^1 \\
\sum x^2 & \sum x^1 & \sum x^0
\end{bmatrix}
\begin{bmatrix} a_2 \\ a_1 \\ a_0 \end{bmatrix}
=
\begin{bmatrix}
\sum x^2 y \\
\sum x^1 y \\
\sum x^0 y
\end{bmatrix}
$$

Calculamos primeiro as somatórias das potências de $x$, que formam a **matriz de equações** (lado esquerdo).

```python
#range(start, end, step)
#print(list(range(2 * g, -1, -1)))

lista_potencias = []

for i in range(2 * g, -1, -1):
    print(((x ** i).sum()))
    lista_potencias.append((x ** i).sum())
```

```python
matriz_esquerda = np.zeros(shape = (g + 1, g + 1), dtype = np.float64)
for i in range(0, g + 1, 1):
    j = g + 1 + i
    trecho = lista_potencias[i:j]
    matriz_esquerda[i] = trecho
display(matriz_esquerda)
```

## Construindo a matriz direita

O vetor $D$ tem $(g+1)$ entradas, onde a $i$-ésima entrada é $\sum x^{g-i} y$.

O loop percorre `i` de `g` até `0` (decrescente), calculando cada somatória cruzada.

```python
matriz_direita = np.zeros(shape = (g + 1, 1), dtype = np.float64)
for i in range(g, -1, -1):
    trecho_x = x ** i
    trecho_y = y
    print((trecho_x * trecho_y).sum())
    matriz_direita[g - i] = (trecho_x * trecho_y).sum()
display(matriz_direita)
```

```python
coeficientes = np.linalg.solve(matriz_esquerda, matriz_direita)
display(coeficientes)
```

```python
#funcao pronta

def mmq(dados: np.ndarray, g: int) -> np.ndarray:

    if not isinstance(dados, np.ndarray):
        raise TypeError("o parametro 'dados' deve ser um array")

    if not isinstance(g, int):
        raise TypeError("o parametro 'g' deve ser um int")
    
    x = dados[:, 0] # slice (fatiamento) do array
    y = dados[:, 1]

    lista_potencias = []
    for i in range(2 * g, -1, -1):
        lista_potencias.append((x ** i).sum())

    matriz_e = np.zeros(shape=(g+1, g+1), dtype=np.float64)
    for i in range(0, g + 1, 1):
        j = g + 1 + i
        trecho = lista_potencias[i:j]
        matriz_e[i] = trecho
    
    matriz_d = np.zeros(shape=(g+1, 1), dtype=np.float64)
    for i in range(g, -1, -1):
        trecho_x = x ** i
        trecho_y = y
        matriz_d[g-i] = (trecho_x * trecho_y).sum()
    
    coeficientes = np.linalg.solve(matriz_e, matriz_d)
    return coeficientes
```

```python
print(mmq(dados, g=2))
```
