```python
import numpy as np
import pandas as pd
import pulp as plp #biblioteca de programacao linear
import matplotlib.pyplot as plt
```

```python
#lendo a base de dados que montamos do nan
df_nan = pd.read_excel("df_nan.xlsx")
display(df_nan)
```

```python
#garantindo que a coluna de horarios ta no formato de hora 
display(pd.to_datetime(df_nan["Hora Início"]))
```

```python
#Trocando as colunas de horas pro horario correto

df_nan["Hora Início"] = pd.to_datetime(df_nan["Hora Início"])

df_nan["Hora Fim"] = pd.to_datetime(df_nan["Hora Fim"])

display(df_nan)
```

```python
df_nan["Hora Início"] = \
    df_nan["Hora Início"].dt.strftime("%H:%M")

df_nan["Hora Fim"] = \
    df_nan["Hora Fim"].dt.strftime("%H:%M")

display(df_nan)
```

```python
df_nad = pd.read_excel("df_nad.xlsx")
display(df_nad)
```

```python
from itertools import product
```

```python
regimes = df_nad["regime"].to_dict()
inicios = df_nan["Hora Início"].to_dict()
print(regimes)
display(inicios)
```

```python
#vamos usar o itertools pra fazer a conta um com todos e todos com um 
#mas poderiamos fazer a mesma coisa usando 2 for, representando as somatorias
#igual na matematica 

lista_variaveis = list(product(regimes, inicios))
display(lista_variaveis)

```

```python
#Criando o problema
prob = plp.LpProblem(name = "Attentus", sense = plp.LpMinimize)

display(prob)
```

```python
dict_variaveis = plp.LpVariable.dicts(name = "A", indices = lista_variaveis, lowBound = 0, cat = plp.LpInteger)
```

```python
lista_coeficientes = []

for indice_regime, indice_periodo in lista_variaveis:
    encargo = df_nad.loc[indice_regime, "enc"]
    lista_coeficientes.append(encargo)

print(lista_coeficientes[::50])
```

```python
lista_multiplicacoes = []

for coef, lpvariable in zip(lista_coeficientes, dict_variaveis.values()):
    lista_multiplicacoes.append(coef * lpvariable)

fo = sum(lista_multiplicacoes)
display(fo)
```
