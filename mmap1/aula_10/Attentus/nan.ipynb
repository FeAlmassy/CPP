```python
coefs = np.polyfit(x = df_duracoes["Percentagem Acumulada"], y = df_duracoes["Duração Conexão"], deg = 5 )

ft = np.poly1d(coefs, variable = "p") #transforma isso numa funcao matematica
print(ft)
```

```python
#para p = 0 => t = -0.20s
print(ft(0))
```

```python
#para p = 1 => t = 37.63s
print(ft(1))
```

```python
np.random.seed(1)
coluna_nan = []
espera = 10
for qtd_ligantes in df_tabela_ligantes["Número Médio Ligantes"]:
    #print(qtd_ligantes)
    momentos_ligacoes = np.random.randint(low = 0, high = 900, size = qtd_ligantes)
    momentos_ligacoes.sort()
    atendentes = []
    for momento_ligacao in momentos_ligacoes:
        #print(momento_ligacao, end = " ")
    #print("-" * 30)
        duracao = ft(np.random.random())

        if len(atendentes) == 0:
            atendentes.append({"inicio" : momento_ligacao,
                              "final"  : momento_ligacao + duracao})
            
            continue

        for atendente in atendentes:
            ocupado = atendente["final"] > momento_ligacao + espera
            if not ocupado:
                
                if momento_ligacao > atendente["final"]:
                    atendente["inicio"] = momento_ligacao
                    atendente["final"] = momento_ligacao + duracao
                else:
                    atendente["inicio"] = atendente["final"]
                    atendente["final"] = atendente["inicio"] + duracao
                break
        else:
            atendentes.append({"inicio" : momento_ligacao,
                              "final"  : momento_ligacao + duracao})  
            
    nan = len(atendentes)
    coluna_nan.append(nan)
print(coluna_nan)
```
