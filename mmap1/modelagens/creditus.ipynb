```python
import pandas as pd
import numpy as np

df  = pd.read_excel("basedados.xlsx")
```

# Etapa 1 - Coleta dos dados

```python
#pra pegar somente uma parte do planilha usando os indices das linhas e colunas 
#vamos usar o iloc

# df.iloc[linhas, coluna]
# df.iloc[primeira_linhas : ultima_linha  ,  primeira_coluna : ultima_coluna]
#o pandas segue o inclusive exclusive do python

#pegando a matrix X (dados dos clientes)
dataframe_dados_clientes = df.iloc[: , 1:19] #pega todas as linhas da coluna da 1 a 18
dataframe_gabarito = df.iloc[: , 19] #pega todas as linhas da coluna 20 (gabarito)

display(dataframe_dados_clientes)
display(dataframe_gabarito)
```

# Etapa 3 - Transformar ambos os dataframe em matrizes

```python
print(type(dataframe_dados_clientes))
print(type(dataframe_gabarito))

#como da pra ver ambos sao do tipo dataframe, pra fazer contas matematicas
#precisamos transformar eles em arrays do numpy

#para fazer isso usaremos o metodo .values do pandas
array_dados_clientes = dataframe_dados_clientes.values
array_gabarito = dataframe_gabarito.values

print(type(array_dados_clientes))
print(type(array_gabarito))
```

# Etapa 4 - Criacao dos cromossomos

```python
def criar_cromossomo(tamanho=19) -> np.array:  #usar tamanho 19 como padrao
    lista_cromossomos = []
    genes = (-1 + 2 * np.random.rand(tamanho)) #lista dos 19 genes
    lista_cromossomos.append(genes)
    vetor_cromossomo = np.array(lista_cromossomos)
    return vetor_cromossomo
```
