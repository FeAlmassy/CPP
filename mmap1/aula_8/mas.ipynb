```python
import numpy as np
```

```python
#funcao pra calcular a raiz de maneira gorila
f = lambda xa: (xa/2) + 1
chute_inicial = 10
xp = f(chute_inicial) #chute inicial
print(xp)
xp = f(xp) #primeira iteracao
print(xp)
xp = f(xp) #segunda iteracao
print(xp)
xp = f(xp) #terceira iteracao
print(xp)
xp = f(xp) #quarta iteracao
print(xp)
```

```python
#primeiro ideia de uma funcao    
f = lambda xa: (xa/2) + 1
chute_inicial = 10
xp = f(chute_inicial)

i = 0
while i < 100:
    xp = f(xp)
    print(xp)
    i += 1
```

```python
#criando uma funcao
def mas(funcao, chute_inicial, qtd_iteracoes):
    xp = funcao(chute_inicial)
    i = 0
    while i < qtd_iteracoes:
        xp = funcao(xp)
        print(xp)
        i += 1

    return xp
```

```python
#melhorando a funcao
def mas(funcao, chute_inicial, qtd_iteracoes, precisao=1e-5):
    xa = funcao(chute_inicial)
    i = 0
    while i < qtd_iteracoes:
        xp = funcao(xa)
        if np.isclose(xp-xa, 0, atol=precisao):
            print(f"Convergiu em {i+1} iterações")
            break
        else:
            xa = xp
            print(f"iteracao {i+1}: {xp}")
        i += 1

    return xp
```

```python
mas(f, chute_inicial, 100)
```
