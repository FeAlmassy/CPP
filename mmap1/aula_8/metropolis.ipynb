```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
import random

#vamos desenhar a imagem com base em funcoes (podem ser varias) e botar isso dentro de um retangulo

r = 1500
qtd_pontos = 1000000
tela = np.ones((2*r,2*r, 3), dtype=np.uint8) * 255
plt.imshow(tela)

cv2.circle(tela, (r,r), r, (0,0,0), 2)
plt.imshow(tela)

def fun_superior(x):
    return np.sqrt(r**2 - (x-r)**2) + r

def fun_inferior(x):
    return -np.sqrt(r**2 - (x-r)**2) + r

i = 0
random.seed(2)
for it in range(qtd_pontos):
    x = random.random() * 2 * r
    y = random.random() * 2 * r
   
    if fun_inferior(x) <= y <= fun_superior(x):
        cv2.circle(tela, (int(x), int(y)), 30, (0,255,0), -1)
        i += 1
    else:    
        cv2.circle(tela, (int(x), int(y)), 30, (0,0,255), -1)

e = qtd_pontos - i
print(f"Quantidade de pontos dentro do circulo: {i}")
print(f"Quantidade de pontos fora do circulo: {e}")
plt.imshow(tela)

```
