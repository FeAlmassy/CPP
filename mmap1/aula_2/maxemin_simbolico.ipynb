Dada uma funcao qualquer de uma entrada, achar os pontos de maximo e minimo de maneira simbolica

### V1 - Código específico

```python
import sympy 
from sympy import *

#condicoes de contorno
x = symbols("x")

f = x**3 + 2*x**2 - 5*x
display(f)
 
#========================

#achando a primeira derivada
d1 = diff(f)
display(d1)

#achando os extremos em forma de lista
extremos = solve(d1, x)
display(extremos)

#achando a segunda derivada
d2 = diff(d1)
display(d2)

for extremo in extremos:
    sinal = d2.subs(x, extremo)
    y = f.subs(x, extremo).evalf()        # y como float
    extremo_num = extremo.evalf()         # x como float
    if sinal > 0:
        print(f"mínimo em (round{extremo_num}, {y})")
    elif sinal < 0:
        print(f"máximo em ({extremo_num}, {y})")
    else:
        print(f"inconclusivo em ({extremo_num}, {y})")


```

### V2 - Função específica

```python
from sympy import *

x = symbols('x')

def extremes_finder(f):
    d1 = diff(f, x)
    d2 = diff(d1, x)
    extremes = solve(d1, x)

    for extreme in extremes:
        signal = d2.subs(x, extreme)
        y = f.subs(x, extreme).evalf()
        extreme_num = extreme.evalf()

        if signal > 0:
            print(f"mínimo em ({extreme_num}, {y})")
        elif signal < 0:
            print(f"máximo em ({extreme_num}, {y})")
        else:
            print("resultado inconclusivo")

extremes_finder(x**3 + 2*x**2 - 5*x)
```

### V3 - Função específica com Grafico

```python
from sympy import *
import plotly.graph_objects as go
import numpy as np

x = symbols('x')

def extremes_finder(f):
    d1 = diff(f, x)
    d2 = diff(d1, x)
    extremes = solve(d1, x)

    # converte a expressão sympy para função numpy
    f_np = lambdify(x, f, 'numpy')

    # range do gráfico
    x_vals = np.linspace(-5, 5, 500)
    y_vals = f_np(x_vals)

    fig = go.Figure()

    # curva da função
    fig.add_trace(go.Scatter(
        x=x_vals, y=y_vals,
        mode='lines',
        name='f(x)',
        line=dict(color='royalblue', width=2)
    ))

    # pontos críticos
    for extreme in extremes:
        signal = d2.subs(x, extreme)
        x_num = float(extreme.evalf())
        y_num = float(f.subs(x, extreme).evalf())

        if signal > 0:
            label = f'mínimo ({x_num:.2f}, {y_num:.2f})'
            color = 'green'
        elif signal < 0:
            label = f'máximo ({x_num:.2f}, {y_num:.2f})'
            color = 'red'
        else:
            label = f'inconclusivo ({x_num:.2f}, {y_num:.2f})'
            color = 'orange'

        fig.add_trace(go.Scatter(
            x=[x_num], y=[y_num],
            mode='markers+text',
            name=label,
            marker=dict(color=color, size=10),
            text=[label],
            textposition='top center'
        ))

    fig.update_layout(
        title='Análise de Extremos',
        xaxis_title='x',
        yaxis_title='f(x)',
        template='plotly_dark'
    )

    fig.show()

extremes_finder(3*x**3 - 8*x**3 + 6*x**2 - x + 5)
```
