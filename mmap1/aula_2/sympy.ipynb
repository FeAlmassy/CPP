```python
from sympy import *
```

```python
#Variaveis
x = symbols('x') #definindo a variavel x
y = symbols('y') #definindo a variavel y
```

```python
#Equacoes
eq1 = Eq(2*x**2 , 4)
display(solve(eq1, x)) #funcao solve isola o x, ou qualquer outra variavel
display(eq1.subs(x, 2)) #metodo subs substitui o valor de
```

```python
#criando um funcao do segundo grau qualquer
f = 2*x**2 + 3*x + 1
display(f) #mostrando a funcao f
#calculando o valor da funcao para x = 2
display(f.subs(x, 2)) #substituindo o valor de x por 2 na funcao f
#calculando a derivada da funcao f em relacao a x
display(diff(f, x)) #derivando a funcao f em relacao a x
#calculando a integral da funcao f em relacao a x
display(integrate(f, x)) #integrando a funcao f em relacao a x
#calculando a integral definida da funcao f em relacao a x, de 0 a 1
display(integrate(f, (x, 0, 1))) #integrando a funcao f em relacao a x, de 0 a 1
```

```python
#criando um funcao do segundo grau qualquer com duas variaveis
g = 2*x**2 + 3*y + 1
display(g) #mostrando a funcao g
#calculando o valor da funcao g para x = 2 e y = 3
display(g.subs({x: 2, y: 3})) #substituindo o valor de x por 2 e y por 3 na funcao g
#calculando a derivada da funcao g em relacao a x
display(diff(g, x)) #derivando a funcao g em relacao a x
#calculando a derivada da funcao g em relacao a y
display(diff(g, y)) #derivando a funcao g em relacao a y
#calculando a integral da funcao g em relacao a x
display(integrate(g, x)) #integrando a funcao g em relacao a
#calculando a integral da funcao g em relacao a y
display(integrate(g, y)) #integrando a funcao g em relacao a y
#calculando a integral definida da funcao g em relacao a x, de 0 a 1
display(integrate(g, (x, 0, 1))) #integrando a funcao g em relacao a x, de 0 a 1
#calculando a integral definida da funcao g em relacao a y, de 0 a 1
display(integrate(g, (y, 0, 1))) #integrando a funcao g em relacao a y, de 0 a 1
```

```python
#Exercicio 1

x = symbols("x")

f = x**3 - 6*x**2 + 9*x + 1
display(f)

der_f_1 = diff(f,x)
display(der_f_1)

extremos = solve(der_f_1, x)
display(extremos)
```
