### Calculadora de Máximo ou Mínimo

Fazer um codigo que percorre uma funcao qualquer em um dominio especifico [a,b] e achar quais sao os pontos de maximo e minimo 

Vamos fazer isso com um for ou um while, comecando o a e indo ate o b, o incremento pra fazer a funcao andar vai ser um dx, escolhido pelo usuario, a cada incremento no dx vai ser usado pra calcular a tangente

pra saber se e ponto de maximo ou minimo nao vamos calcular a segunda derivada, ate pq isso precisaria de muita gambiarra ou metodos simbolicos

pra fazer isso de mandeira burra (computacional) vamos ver que, se o ponto e de maximo as tangentes anteriores vao ser positivas, enquanto que no ponto de minimo as tangentes anteriores vao ser negativas

```python
import math

f = lambda x: 2 * x ** 3 + 2 * x ** 2

lim_inferior = -2
lim_superior = 1
qtd_particoes = 300
dx = (lim_superior - lim_inferior) / qtd_particoes
tolerancia = 0.1
print(f"o dx vale: {dx}")

def derivada(f, x: float, dx: float):
    dy = f(x + dx) - f(x)
    tg = dy / dx
    return tg

i = lim_inferior
extremos = []
derivadas = []

while i < lim_superior:
    d_atual = derivada(f, i, dx)
    derivadas.append([i, d_atual])
    if abs(d_atual) < tolerancia:
        extremos.append((i, d_atual)) 
    i += dx

display(extremos)

maximos = []
minimos = []
extremo_anterior = extremos[0]

for extremo_atual in extremos[1:]:
    d_anterior = extremo_anterior[1]
    d_atual    = extremo_atual[1]

    if d_anterior > 0 and d_atual < 0:
        maximos.append(extremo_atual[0])
    elif d_anterior < 0 and d_atual > 0:
        minimos.append(extremo_atual[0])

    extremo_anterior = extremo_atual

print("Máximos:")
for x in maximos:
    print(f"x = {x:.5f}   f(x) = {f(x):.5f}")

print("Mínimos:")
for x in minimos:
    print(f"  x = {x:.5f}   f(x) = {f(x):.5f}")
```
