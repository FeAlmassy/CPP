```python
import numpy as np
import matplotlib.pyplot as plt
import random
import time

r          = 1500
qtd_pontos = 100_000_000

def fun_superior(x):
    return np.sqrt(r**2 - (x - r)**2) + r

def fun_inferior(x):
    return -np.sqrt(r**2 - (x - r)**2) + r

def dentro_do_circulo(x, y):
    return fun_inferior(x) <= y <= fun_superior(x)

def metropolis(qtd_pontos, r):
    xs_dentro = []
    ys_dentro = []
    xs_fora   = []
    ys_fora   = []
    i = 0

    for _ in range(qtd_pontos):
        x = random.random() * 2 * r
        y = random.random() * 2 * r

        if dentro_do_circulo(x, y):
            xs_dentro.append(x)
            ys_dentro.append(y)
            i += 1
        else:
            xs_fora.append(x)
            ys_fora.append(y)

    return xs_dentro, ys_dentro, xs_fora, ys_fora, i

random.seed(2)
inicio = time.perf_counter()
xs_dentro, ys_dentro, xs_fora, ys_fora, i = metropolis(qtd_pontos, r)
fim = time.perf_counter()

e = qtd_pontos - i
print(f"Quantidade de pontos dentro do círculo: {i}")
print(f"Quantidade de pontos fora do círculo:   {e}")
print(f"Tempo Python: {fim - inicio:.3f}s")

plt.figure(figsize=(7, 7))
plt.scatter(xs_fora,   ys_fora,   color="blue",  s=0.1, alpha=0.2, label="Fora")
plt.scatter(xs_dentro, ys_dentro, color="green", s=0.1, alpha=0.2, label="Dentro")
theta = np.linspace(0, 2 * np.pi, 1_000)
plt.plot(r + r * np.cos(theta), r + r * np.sin(theta), color="black", lw=2, label="Círculo")
plt.gca().set_aspect("equal")
plt.title("Metropolis")
plt.legend()
plt.show()
```
