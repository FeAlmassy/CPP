```python

def bhaskara(a:float, b:float, c:float):
    delta = (b ** 2) - 4 * a * c
    if delta < 0:
        print(f"nao existem raizes reais negativas")
    elif delta > 0:
        x1 = -b + np.sqrt(delta) / 2 * a
        x2 = -b - np.sqrt(delta) / 2 * a
        print(f"existem duas raizes reais distintas, sendo elas: x1={x1} e x2={x2}")
    else:
        print(f"existem duas raizes reais e iguais, sendo ela: x1={x1}")
        

bhaskara(2,4,0)
```
