## 1) Condições (if, else, elif)

```python
# Estrutura geral do bloco IF

# 1a opção
# if [True/False] and/or [True/False] and/or [True or False] ...:
#       código a ser executado caso a condição acima SEJA satisfeita (True)
# elif [True/False] and/or [True/False] and/or [True/False] ...:
#       código a ser executado caso a condição acima SEJA satisfeita (True)
# else:
#       código a ser executado caso a condição NÃO SEJA satisfeita (False)

# 2a opção
### bloco a ser executado if [True/False] and/or [True/False] ... else [bloco a ser executado caso nenhuma condição satisfeita]
```

```python
n = 3
resultado = n > 2
print(f"Resultado: {resultado}")    # f-string
```

```python
# O trecho 'else' só ocorre quando o 'if' não é satisfeito

# Opção 1
# if resultado is True:                             # Outra possibilidade para comparar um valor booleano
if resultado == True:                               # Se (0 indentações)
    print("'n' é maior que 2")                      # Então (1 indentação)
else:                                               # Senão (0 indentações, mesmo nível do 'if')
    print("'n' é menor ou igual a 2")               # Então (1 indentação)

print("Este print é independente da variável 'resultado'")
```

```python
# o bloco 'if' não precisa ter um 'else' e nem 'elif'
# como a variavel 'resultado' é booleana, podemos simplificar o código da seguinte forma (sem o '== True')
if resultado:
    print("'n' é maior que 2")          # NUNCA ESQUEÇA DE INDENTAR O CÓDIGO DENTRO DO BLOCO 'if'

print("Este print não depende da variável 'resulta'")
```

```python
# O trecho 'else' só ocorre quando o 'if' E TODOS os 'elif' não são satisfeitos
# É IMPORTANTE lembrar que a primeira condição que for satisfeita, o programa entra no
# bloco e sai do 'if'
numero = 0
if numero > 5:
    print("O número é maior que 5.")
elif numero > 2:    # (NÃO É NECESSÁRIO HAVER UM ELIF. SOMENTE SE A LÓGICA EXIGIR)
    print("O número é menor/igual a 5 e maior que 2.")
elif numero < 2:    # (NÃO É NECESSÁRIO HAVER UM ELIF. SOMENTE SE A LÓGICA EXIGIR)
    print("O número é menor que 2.")
else:               # (NÃO É NECESSÁRIO HAVER UM ELSE. SOMENTE SE A LÓGICA EXIGIR)
    print("O número é 2.")
```

```python
# Os trechos abaixo são denominados de 'if aninhado (nested)'
# Mostrar o debugger
n = 20
if n > 10:
    if n > 20:
        print("'n' é maior que 20")
    else:
        print("'n' é menor ou igual a 20 e maior que 10")

    print("Acabou o 'if' principal")
else:
    if n > 5:
        print("'n' é maior que 5 e menor ou igual a 10")
    else:
        print("'n' é menor ou igual a 5")

    print("Acabou o 'else' principal")

print("Acabou todos os 'ifs'")
```

##### Perceber que blocos de **condição** (``if``, ``else`` ou ``elif``) operam apenas sobre dados booleanos

#### Os seguintes valores são considerados 'falsey', ou seja, eles são considerados 'False' quando aplicados em blocos de condições.

* 0, or any numerical value equivalent to zero (0, ou algum valor numérico equivalente a zero)
* Empty sequences (Sequências vazias): '', "", (), []
* All other values in Python evaluate to True. (Todos os outros valores em Python são avaliados como True.)

## 2) Repetições (while e for)

```python
# Para estudar laços de repetições é crucial entender o conceito de 'incremento' e
# 'decremento'
# Incremento: adicionar unidade(s)
# Decremento: subtrair unidade(s)
# Exemplo:

# anotando a variavel 'n' como int. Porém o python não vai forçar que 'n' aceite apenas
# inteiros anotações de variáveis em python servem apenas para documentação
n = 12

# Incrementos mais comuns
n = n + 1      # (incremento) Retorna 13
print(n)
n = n - 1      # (decremento) Retorna 12
print(n)
```

```python
# Incrementos menos comuns
n = n * 2      # (multiplicação) Retorna 24
print(n)
n = n / 2      # (divisão) Retorna 12
print(n)
n = n ** 2     # (potenciação) Retorna 144
print(n)
n = n // 2     # (divisão inteira) Retorna 0
print(n)
n = n % 2      # (resto da divisão) Retorna 0
print(n)
```

```python
i = 0

# laço de repetição "burro"...
print(i)
i = i + 1
print(i)
i = i + 1
print(i)
i = i + 1
print(i)
i = i + 1
print(i)
i = i + 1
print(i)
i = i + 1
print(i)
i = i + 1
print(i)
```

### Os laços de repetição **'while'** exige uma condição (True/False) para ser executado

```python
# mostrar pelo debugger
i = 0           # i é chamado de 'variável de controle'
while i <= 7:   # Enquanto i <= 7: (faça)
    print(i)
    i = i + 1

print("fim do programa")
```

```python
# outro laço de repetição burro...
print(0)
print(1)
print("Pulando o 2...")
print(3)
print("Saindo do while antes de imprimir o 4")
print("while terminou!!!")
```

```python
# Percorrendo um laço 'while' com 'break' e 'continue'
i = 0
while i < 7:
    if i == 2:
        print("Pulando o 2...")
        i = i + 1
        continue

    if i == 4:
        print("Saindo do while antes de imprimir o 4")
        break

    print(i)
    i = i + 1

print("While terminou!!!")
```

# 1) Imprimir Números de 1 a 10:
#### Use um loop while para imprimir números de 1 a 10.

```python
# Inicializa a variável
i = 1
# Loop enquanto i for menor ou igual a 10
while i <= 10:
    print(i)
    i += 1  # Incrementa i
```

# 2) Contagem Regressiva de 10 a 1:
#### Use um loop while para imprimir números de 10 a 1.

```python
i = 10
while i >= 1:
    print(i)
    i -= 1  # Decrementa i
```

# 3) Soma dos Primeiros N Números Naturais:
#### Peça ao usuário um número n e calcule a soma dos primeiros n números naturais usando um loop while.

```python
n = int(input("Digite um número: "))
soma = 0
i = 1
while i <= n:
    soma += i  # Adiciona i à soma
    i += 1
print("A soma é:", soma)
```

# 4) Imprimir Números Pares:
#### Use um loop while para imprimir todos os números pares entre 1 e 20.

```python
i = 1
while i <= 20:
    if i % 2 == 0:  # Verifica se i é par
        print(i)
    i += 1
```

# 5) Tabuada de Multiplicação:
#### Peça ao usuário um número e imprima sua tabuada até 10 usando um loop while.

```python
num = int(input("Digite um número: "))
i = 1
while i <= 10:
    print(f"{num} x {i} = {num * i}")
    i += 1
```

# 8) Sequência de Fibonacci:
#### Use um loop while para montar uma lista com os primeiros 10 n números da sequência de Fibonacci. No final, imprima a lista finall

```python
sequencia_fibonacci = [0, 1]
i = 2
while i < 10:
    sequencia_fibonacci.append(sequencia_fibonacci[i - 1] + sequencia_fibonacci[i - 2])
    i += 1

print(sequencia_fibonacci)
```

# 9) Soma de Números Digitados pelo Usuário:
#### Peça ao usuário para digitar números continuamente. Pare quando o usuário digitar 0 e mostre a soma dos números digitados.

```python
soma = 0
while True:
    num = int(input("Digite um número (0 para parar): "))
    if num == 0:
        break
    soma += num

print("Soma dos números digitados:", soma)
```

# 10) Contar Vogais em uma String:
#### Peça ao usuário uma palavra e use um loop while para contar quantas vogais (a, e, i, o, u) ela contém.

```python
palavra = input("Digite uma palavra (apenas minusculos): ")
contador = 0
i = 0
while i < len(palavra):
    if palavra[i] == "a" or palavra[i] == "e" or palavra[i] == "i" or palavra[i] == "o" or palavra[i] == "u":
        contador += 1
    i += 1
print("Número de vogais:", contador)
```

# 11) Adivinhação de senhas:
#### Crie uma senha e monte um while loop para permitir ao usuário (input) tentar adivinhar a senha.
#### após tres tentativas mal sucedidas, imprima uma mensagem de 'cartão bloqueado'

```python
senha = "1234"
tentativas = 0
while tentativas < 3:
    entrada = input("Digite a senha: ")
    if entrada == senha:
        print("Acesso permitido")
        break
    else:
        print("Senha incorreta. Tente novamente")

    if tentativas == 2:
        print("Cartão bloqueado")

    tentativas += 1
```
