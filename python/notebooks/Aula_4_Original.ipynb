```python
# criando uma função para encontrar a periodicidade de um número fracionário
# o parametro (pf_string) deve ser uma string que representa um número fracionário
# este parâmetro é obrigatório, pois não possui um valor padrão definido.
# A função retorna um valor inteiro que representa a periodicidade do número fracionário
def encontra_periodicidade(pf_string: str) -> int:
    """
    Função que encontra a periodicidade de um número fracionário
    :param pf: float: Número fracionário
    
    :return: int: 
    Periodicidade do número fracionário. Por exemplo:
    0.3333 -> p = 1
    0.121212 -> p = 2
    0.123123 -> p = 3
    """

    # inicia a variável 'p' com 1, pois a primeira periodicidade possível é 1
    p = 1

    # devido ao fato de que o número fracionário é uma string, o primeiro índice
    # da parte fracionária é 2
    x = 2
    while True:

        y = x + p
        z = y + p

        trecho_um = pf_string[x:y]
        trecho_dois = pf_string[y:z]

        # se os trechos forem iguais, então a periodicidade foi encontrada
        if trecho_um == trecho_dois:
            break
            
        p += 1
    
    return p
```

```python
# criando uma função para converter um número fracionário periódico em uma fração
# o parâmetro (numero_float) é um número fracionário que será convertido em uma fração
# este parâmetro é opcional, pois possui um valor padrão definido
# A função retorna uma string que representa o número fracionário convertido em uma 'fração'
def main(numero_float: float = 3.1414) -> str:

    """
    Função que converte um número fracionário periódico em uma fração
    :param pf: float: Número fracionário

    :return: str:
    Número fracionário convertido em fração. Por exemplo:
    0.3333 -> 1/3
    0.121212 -> 12/99
    0.123123 -> 123/999
    """

    # convertendo o número fracionário em uma string
    numero_string = str(numero_float)

    # separando a parte inteira da parte fracionária
    pi, pf = numero_string.split(".")
    
    # convertendo a string 'pi' em um número inteiro
    num_pi = int(pi)

    # o denominador da fração que representa a parte inteira (pi) é sempre 1
    den_pi = 1
    
    # fazendo com que 1414 (pf) se  converta em um float no tipo 0.1414
    pf_float = float(f"0.{pf}")
    pf_string = str(pf_float)

    # chamando a função auxiliar para encontrar a periodicidade do número fracionário
    p = encontra_periodicidade(pf_string)

    # encontrando o numerador da fração que representa a parte fracionária (pf)
    num_pf = int(pf_float * 10**p)

    # encontrando o denominador da fração que representa a parte fracionária (pf)
    den_pf = 10**p - 1

    # encontrando o numerador e o denominador da fração final
    numerador_final = (num_pi * den_pf + num_pf)

    # encontrando o denominador da fração final
    denominador_final = den_pf

    # montando a 'fração' final
    fracao_final = f"{numerador_final}/{denominador_final}"
    return fracao_final
```

```python
# chamando a função principal com parametro posicional
print(main(0.3333))                 # parametro posicional

# chamando a função principal com parametro nomeado
print(main(numero_float=0.3333))    # parametro nomeado
```

```python
# conhecendo enumerate()
# enumerate(iterable, start)
# iterable: objeto iterável (lista, tupla, dicionário, etc) OBRIGATÓRIO
# start: valor inicial (padrao = 0) OPCIONAL

# Exemplo:
# enumerate([10, 20, 30], 0) = (0, 10), (1, 20), (2, 30)
# enumerate([10, 20, 30], 1) = (1, 10), (2, 20), (3, 30)
# enumerate([10, 20, 30], 2) = (2, 10), (3, 20), (4, 30)

# enumerate(nomes) = enumerate(nomes, 0)
# enumerate(nomes) = (0, "João"), (1, "Maria"), (2, "José")

for par in enumerate(nomes):
    # separando o índice e o valor de dentro da variável 'par'
    indice, nome = par
    print(f"indice: {indice}, nome: {nome}")

print("-" * 50)

# Fazer na hora. Primeiro, sem o enumerate e depois com o enumerate
for indice, valor in enumerate(nomes):
    print(f"indice: {indice}, valor: {valor}")
```

```python
# enumerate(nomes, 0) = (0, "João"), (1, "Maria"), (2, "José")
# enumerate(nomes, 1) = (1, "João"), (2, "Maria"), (3, "José")
# enumerate(nomes, 2) = (2, "João"), (3, "Maria"), (4, "José")

for par in enumerate(nomes, 1):
    # separando o índice e o valor de dentro da variável 'par'
    indice, nome = par
    print(f"indice: {indice}, nome: {nome}")

print("-" * 50)

# enumerate(nomes) = enumerate(nomes, 0)
for indice, nome in enumerate(nomes, 2):
    print(f"indice: {indice}, nome: {nome}")
```

```python
# conhecendo zip()
# zip(iterables)
# iterables: objetos iteráveis (lista, tupla, dicionário, etc)

# Exemplo:
# zip([10, 20, 30], [40, 50, 60]) = (10, 40), (20, 50), (30, 60)
# zip([10, 20, 30], [40, 50, 60], [70, 80, 90]) = (10, 40, 70), (20, 50, 80), (30, 60, 90)

nomes = ["João", "Maria", "José"]
nomes_meio = ["da", "de", "do"]
sobrenomes = ["Silva", "Souza", "Santos"]
# zip(nomes, sobrenomes) = ("João", "Silva"), ("Maria", "Souza"), ("José", "Santos")

for trio in zip(nomes, nomes_meio, sobrenomes):
    nome, nome_meio, sobrenome = trio
    print(f"nome completo: {nome} {nome_meio} {sobrenome}")

print("-" * 50)

for nome, nome_meio, sobrenome in zip(nomes, nomes_meio, sobrenomes):
    print(f"nome completo: {nome} {nome_meio} {sobrenome}")
```
