# `Operadores`
## Cada objeto possui um comportamento específico para cada operador. Os operadores podem ser **aritméticos** ou de **comparação**

````python
# operadores aritméticos para int e/ou float
print(10 + 5)       # soma
print(10 - 5)       # subtração
print(10 * 5)       # multiplicação
print(10 / 5)       # divisão
print(10 // 5)      # divisão inteira
print(10 % 5)       # resto da divisão (mod)
print(10 ** 5)      # potência
````

````python
# operadores aritméticos para str
print("Python" + " é " + "legal")      # concatenação
print("Python" * 3)                     # repetição
````

````python
# operadores relacionais (comparação) para int e/ou float
print(10 > 5)       # maior que
print(10 < 5)       # menor que
print(10 >= 5)      # maior ou igual
print(10 <= 5)      # menor ou igual
print(10 == 5)      # igual
print(10 != 5)      # diferente
````

````python
# operadores lógicos para bool
print(True and True)    # e
print(True or False)    # ou
print(not False)        # não
````

### Cada objeto possui, também, seus **métodos**. Esses métodos são **intrinsecos** aos objetos. Observe os exemplos:

````python
# métodos para str
print("igor".startswith("i"))               # retorna True pois a string começa com a letra 'i'
print("igor".endswith("r"))                 # retorna True pois a string termina com a letra 'r'
print("igor".upper())                       # retorna a string em maiúsculo
````

# `listas`
#### O tipo list é provavelmente o tipo de coleção mais usado em Python. Em Python, uma list é apenas uma coleção **ordenada** de valores. A list deve ser criada com valores separados por VÍRGULA entre COLCHETES. Por exemplo: [1, 3, 4, 2] é uma list contendo 4 valores inteiros. As list são **mutáveis**, o que significa que você pode alterar seus elementos. Você pode acessar, inserir, ou remover elementos facilmente.

### **Slicing** de lista

````python
# subendereços de uma coleção (índices)
# indices  0    1    2    3    4    5    6    7    8
lista1 = ["a", "b", "c", "d", "e", "f", "g", "h", "i"]
# indices -9   -8   -7   -6   -5   -4   -3   -2   -1

lista2 = [1, "a", 2, "b", 3, "c", 4, "d", 5, "e", 6]
lista3 = [1, 2, ["a", "b"]]
````

````python
# todas as regras de acesso (slicing) de uma coleção se aplicam a todas as coleções
print(lista1[-1])
````

````python
print(lista3[-1])
````

````python
# quando há coleção dentro de uma outra coleção, o acesso é feito em duas (ou mais) etapas
# lista3[-1] retorna a lista ["a", "b"]
print(lista3[-1][0])
````

````python
print(lista1[0:5]) # retorna os elementos a partir do índice 0 até o índice 4 (5 - 1)
print(lista1[0:3]) # retorna os elementos a partir do índice 0 até o índice 2 (3 - 1)
print(lista1[3:7]) # retorna os elementos a partir do índice 3 até o índice 6 (7 - 1)
````

````python
# imprimindo a lista de frente para trás um a um
print(lista1[::1])
````

````python
# imprimindo a lista de trás para frente um a um
print(lista1[::-1])
````

````python
# para utilizar o passo como valor negativo, é necessário que o índice inicial seja maior que o índice final
# caso contrário, a lista será retornada vazia
print(lista1[2:4:-1])
````

````python
# vai de 4 até 1 (0 + 1) de um em um na ordem inversa
print(lista1[4:0:-1])
````

````python
# vai de 4 até 1 (0 + 1) de dois em dois na ordem inversa
print(lista1[4:0:-2])
````

````python
# invertendo a lista de trás para frente e imprimindo os elementos a partir do índice 5 até o índice 1
print(lista1[5:1:-1])
print(lista1[-4:-8:-1])
````

### **Métodos** de lista

````python
# list.append - adiciona um elemento ao final da lista
lista1.append("j")
print(lista1)
````

````python
# list.insert - adiciona um elemento em uma posição específica da lista
lista1.insert(0, "z")
print(lista1)
````

````python
# list.insert - adiciona um elemento em uma posição específica da lista
lista1.insert(8, "y")
print(lista1)
````

````python
# lista1[::1] retorna a lista de frente para trás do começo ao fim um a um e apenas IMPRIME
print(lista1[::1])
````

````python
# lista1[::-1] retorna a lista de trás para frente do fim ao começo um a um e atualiza a lista
lista1 = lista1[::-1]
print(lista1)
````

````python
# list.reverse - inverte a ordem dos elementos da lista e atualiza a lista (reatribuição automatica)
lista1.reverse()
print(lista1)
````

````python
# list.pop - remove o elemento presente no índice específicado
lista1.pop(0)
print(lista1)
````

````python
# o método pop retorna o elemento removido depois de removê-lo da lista e IMPRIME
print(lista1.pop(0))
````

````python
print(letra_jogada_fora, lista1)
````

As tuplas são comumente usadas para pequenas coleções de valores que não precisarão mudar, como um endereço IP e uma porta. As tuplas são representadas com **parênteses** em vez de colchetes

````python
# indices   0  1  2  3  4  5  6  7  8
tupla_um = (1, 2, 3, 4, 5, 6, 7, 8, 9)
# indices  -9 -8 -7 -6 -5 -4 -3 -2 -1

tupla_dois = ("a", "b", "c", "d", "e", "f", "g")
tupla_tres = (1, 2, ["a", "b"], ("x", "y"))
````

# `Dicionários`
### Um dicionário em Python é uma coleção de pares de **chave-valor**. O dicionário é cercado por **chaves**. Cada par é separado por uma vírgula e a chave e o valor são separados por **dois pontos**

````python
estados_capitais = {'Pernambuco': 'Recife',
                    'Goias': 'Goiânia',
                    'Rio Grande do Sul': 'Porto Alegre',
                    'Paraná': 'Curitiba'
                    }
````

# `f-strings`
### f-strings são o mesmo que strings, porém possui um recurso para deixar a string formatável, e portanto, mais inteligente

````python
a = 1
b = 2

# dentro das chaves são valores não-literais (variáveis) e fora das chaves são valores literais
print(f"a: {a} - b: {b}")

a, b = b, a

print(f"a: {a} - b: {b}")
````

### Formatando strings com números...

````python
# f"texto literal {numero:formato}"
# os formatos podem ser:
# s - string
# d - decimal
# f - float
# x - hexadecimal
# b - binário
# e - notação científica
# % - porcentagem
````

````python
# formatando float...
variavel = 322.563
print(f"valor: {variavel:015.5f}")
print(f"valor: {variavel:.1f}")
print(f"valor: {variavel:.2f}")
print(f"valor: {variavel:.3f}")
print(f"valor: {variavel:.4f}")
````

````python
# formatando porcentagem...
percentual = 0.25
print(f"percentual: {percentual:%}")
print(f"percentual: {percentual:010.0%}")
print(f"percentual: {percentual * 100}%")   # Método Cavalo
print(f"percentual: {percentual:.0%}")      # Método Intelogente
print(f"percentual: {percentual:.1%}")
print(f"percentual: {percentual:.2%}")
````

````python
# formatando inteiro...
inteiro = 20
print(f"inteiro: {inteiro:01d}")
print(f"inteiro: {inteiro:02d}")
print(f"inteiro: {inteiro:03d}")
print(f"inteiro: {inteiro:04d}")
````

````python
# formatando string...
texto = "Python"
print(f"texto: {texto:10s}")            # alinhamento à esquerda (padrão)
print(f"texto: {texto:>10s}")           # alinhamento à direita
print(f"texto: {texto:^10s}")           # centralizado
print(f"texto: {texto:<10s}")           # alinhamento à esquerda
````

````python
# formatando hexadecimal...
inteiro = 255
print(f"inteiro: {inteiro:08x}")          # hexadecimal minúsculo
print(f"inteiro: {inteiro:8X}")          # hexadecimal maiúsculo
````

````python
# formatando binário...
inteiro = 255
print(f"inteiro: {inteiro:32b}")          # binário
print(f"inteiro: {inteiro:032b}")          # binário
````

````python
# formantando para notação científica...
inteiro = 1_000_000                     # _ dentro de um número é ignorado (pode servir como separador de milhar para facilitar a leitura)
print(f"inteiro: {inteiro:.1e}")
print(f"inteiro: {inteiro:.10e}")          # notação científica minúscula
print(f"inteiro: {inteiro:.8E}")          # notação científica maiúscula
````

# `Blocos de Condição`
### Os blocos de condição servem para executarem (ou não) comandos **condicionalmente**. Não podemos esquecer que abaixo de um bloco de condição devemos usar uma **indentação (TAB)**

````python
n = 3
resultado = n > 2                                   # resultado recebe True se n for maior que 2, senão recebe False

# Opção 1 de if-else
if resultado == True:                               # Se (0 indentações)
    print("'n' é maior que 2")                      # Então (1 indentação)
else:                                               # Senão (0 indentações, mesmo nível do 'if')
    print("'n' é menor ou igual a 2")               # Então (1 indentação)

# Opção 2 de if-else (sem indentação)
# Não recomendado para iniciantes. Só incluí para termos de documentação para futuras consultas
print("'n' é maior que 2") if resultado == True else print("'n' é menor ou igual a 2")

# devido à indentação, o print abaixo está fora do bloco do if-else
print("final do print")
````

````python
# CUIDADO COM A INDENTAÇÃO...

numero = 2
if numero > 5:
    print("O número é maior que 5.")
elif numero > 2:    # (NÃO É NECESSÁRIO HAVER UM ELIF. SOMENTE SE A LÓGICA EXIGIR)
    print("O número é menor/igual a 5 e maior que 2.")
elif numero < 2:    # (NÃO É NECESSÁRIO HAVER UM ELIF. SOMENTE SE A LÓGICA EXIGIR)
    print("O número é menor que 2.")
else:               # (NÃO É NECESSÁRIO HAVER UM ELSE. SOMENTE SE A LÓGICA EXIGIR)
    print("O número é 2.")
````
