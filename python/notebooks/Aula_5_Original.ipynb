```python
# criando uma função para gerar a tabuada
def gera_tabuada(i_inicial, i_final, j_inicial, j_final, separador=" | "):

    """
    Função para gerar a tabuada de uma coleção de números

    :param i_inicial: int - Número inicial da coleção de números
    :param i_final: int - Número final da coleção de números
    :param j_inicial: int - Número inicial da coleção de números
    :param j_final: int - Número final da coleção de números
    :param separador: str - Separador entre os números da tabuada. Por padrão é " | "
    """

    colecao_i = range(i_inicial, i_final + 1)
    colecao_j = range(j_inicial, j_final + 1)

    multiplicacao_maxima = colecao_i[-1] * colecao_j[-1]
    qtd_digitos = len(str(multiplicacao_maxima))

    tamanho_separador = len(separador)

    for i in colecao_i:
        comprimento_texto = 0
        for j in colecao_j:
            texto = f'{i:0{qtd_digitos}} x {j:0{qtd_digitos}} = {i * j:0{qtd_digitos}}'
            print(texto, end=separador)
            comprimento_texto = comprimento_texto + len(texto) + tamanho_separador
        
        comprimento_texto = comprimento_texto - 1
        print()
        print("-" * comprimento_texto)

gera_tabuada(1, 15, 5, 10)
```

```python
def simplificacao_binaria(param_tipo_cruzamento, param_cosseno):

    """
    Função para simplificar a condição de registro do evento

    :param param_tipo_cruzamento: str (in, out, both)
    :param param_cosseno: float (-1 <= cosseno <= 1) E cosseno != 0
    """

    if param_tipo_cruzamento == "in" and param_cosseno > 0 or \
       param_tipo_cruzamento == "both" and param_cosseno > 0 or \
       param_tipo_cruzamento == "out" and param_cosseno < 0 or \
       param_tipo_cruzamento == "both" and param_cosseno < 0:
        print("Registrar o evento")
    else:
        print("Não registrar o evento")

    A = param_tipo_cruzamento == "in"
    B = param_tipo_cruzamento == "both"
    C = param_cosseno > 0

    S = (not B) and (not operator.xor(A, C)) or (not A and B)
    if S == True:
        print("Registrar o evento")
    else:
        print("Não registrar o evento")

simplificacao_binaria("out", -0.5)
print("-" * 10)
simplificacao_binaria("out", 0.5)
print("-" * 10)
simplificacao_binaria("both", -0.5)
print("-" * 10)
simplificacao_binaria("both", 0.5)
print("-" * 10)
simplificacao_binaria("in", -0.5)
print("-" * 10)
simplificacao_binaria("in", 0.5)
```

```python
# 'passo' não pode ser um número que não seja múltiplo de 100
# então é necessário uma função para definir os múltiplos de 100
def busca_multiplos(param_valor):
    multiplos = []
    for i in range(1, param_valor // 2 + 1):
        if param_valor % i == 0:
            multiplos.append(i)
    return multiplos

print(busca_multiplos(5))
print(busca_multiplos(10))
print(busca_multiplos(100))
print(busca_multiplos(1000))
```

```python
def monetinha(param_passo: int = 10):
    """
    Função que gera as carteiras de ações com 3 ativos

    :param param_passo int: passo para a variação dos ativos
    :return: lista com as carteiras de ações
    """

    multiplos_cem = busca_multiplos(100)
    if param_passo not in multiplos_cem:
        raise ValueError("O passo deve ser um múltiplo de 100")
    
    if not isinstance(param_passo, int):
        raise TypeError("O passo deve ser um número inteiro")

    carteiras = []

    colecao_petr = list(range(0, 100 + 1, param_passo))

    for perc_petr in colecao_petr:
        colecao_embr = list(range(0, 100 - perc_petr + 1, param_passo))
        for perc_embr in colecao_embr:
            perc_vale = 100 - perc_petr - perc_embr
            carteira = [perc_petr, perc_embr, perc_vale]
            carteiras.append(carteira)
    
    return carteiras

print(f"{len(monetinha(param_passo=1))} carteiras")
print(f"{len(monetinha(param_passo=5))} carteiras")
print(f"{len(monetinha(param_passo=10))} carteiras")
print(f"{len(monetinha(param_passo=25))} carteiras")
```

```python
def mini_hurst(param_serie: list | tuple):

    """
    Função que calcula o mini-Hurst de uma série temporal

    :param param_serie: lista ou tupla com a série temporal

    :raises TypeError: se a série não for uma lista ou uma tupla
    :raises ValueError: se a quantidade de elementos da série não for 
    uma potência de 2
    """
    
    if not isinstance(param_serie, (list, tuple)):
        raise TypeError("A série deve ser uma lista ou uma tupla")

    qtd_elementos = len(param_serie)
    
    # checar se a quantidade de elementos é uma potência de 2
    log2 = math.log2(qtd_elementos)
    if log2 % 1 != 0:
        raise ValueError("A quantidade de elementos da série não é uma potência de 2")
    

    i = 1
    while i <= math.log2(len(param_serie)):
        trecho = param_serie[0:2 ** i]

        media = statistics.mean(trecho)
        desvio_padrao = statistics.stdev(trecho)
        print(f"Trecho: {trecho} | Média: {media:.2f} | Desvio padrão: {desvio_padrao:.2f}")

        i = i + 1

mini_hurst(serie)
```

```python
def encontra_otimo(funcao, pontos: list[tuple | list], 
                   tipo: str = "max") -> tuple:

    """
    esta funcao serve para encontrar o ponto otimo (maximo ou minimo) de 
    uma funcao

    :param funcao: função - função a ser otimizada
    :param pontos: list - lista de pontos para testar
    :param tipo: str - tipo de otimização (max ou min)

    :return: tuple - melhor ponto e valor da função objetivo
    """

    ponto_testar = pontos[0]
    valor_fo = funcao(*ponto_testar)

    melhor_ponto = ponto_testar
    melhor_valor = valor_fo

    for ponto in pontos[1:]:
        valor_fo = funcao(*ponto)
        if tipo == "max" and valor_fo > melhor_valor or \
           tipo == "min" and valor_fo < melhor_valor:
            melhor_valor = valor_fo
            melhor_ponto = ponto

    return (melhor_ponto, melhor_valor)

ponto_otimo, valor_otimo = encontra_otimo(fo, pontos_maximo, "max")
print(f"Melhor ponto: {ponto_otimo} com valor de {valor_otimo}")
```

```python
import random
```

```python
def quadrado(param_numero):
    return param_numero ** 2
```

```python
def encontra_vetores(param_qtd_dimensoes: int = 10) -> list:

    """
    Função para gerar vetores aleatórios

    :param param_qtd: int - quantidade de vetores a serem gerados
    :return: list - lista com os vetores gerados
    """

    vetor_1 = []
    vetor_2 = []

    # gerando os vetores aleatoriamente
    i = 0
    while i < param_qtd_dimensoes:
        vetor_1.append(random.randint(-10, 10))
        vetor_2.append(random.randint(-10, 10))

        i = i + 1

    # calculando o produto escalar
    produto_escalar = 0
    i = 0
    while i < param_qtd_dimensoes:
        produto_escalar = produto_escalar + vetor_1[i] * vetor_2[i]
        i = i + 1

    # calculando as normas
    norma_1 = sum(map(quadrado, vetor_1)) ** 0.5
    norma_2 = sum(map(quadrado, vetor_2)) ** 0.5

    # tratando uma possível situação de divisão por zero
    try:
        # calculando o cosseno
        cosseno = produto_escalar / (norma_1 * norma_2)
    except ZeroDivisionError:
        return None, None

    # tratando uma possível situação de cosseno maior que 
    # 1 (1.00000001) ou menor que -1 (-1.00000001)
    try:
        # calculando o ângulo em radianos
        angulo_rad = math.acos(cosseno)
    except ValueError:
        if cosseno > 1:
            angulo_rad = 0
        elif cosseno < -1:
            angulo_rad = math.pi

    # calculando o ângulo em graus
    angulo_graus = math.degrees(angulo_rad)

    return angulo_graus, (vetor_1, vetor_2)
```

```python
def otimizacao_natural(funcao_otimizar, valor_soma: int | float, 
                       qtd_niveis: int = 3, 
                       dominio: np.ndarray = np.arange(0, 5, 1), 
                       tipo: str = "max"):

    """
    Função para otimização natural

    :param funcao_otimizar: função - função a ser otimizada
    :param valor_soma: int - valor da soma dos parâmetros
    :param qtd_niveis: int - quantidade de níveis
    :param dominio: np.ndarray - domínio dos parâmetros
    :param tipo: str - tipo de otimização (max ou min)

    :return: tuple - melhor ponto e valor da função objetivo
    """

    # checando se o tipo é válido
    if tipo not in ["max", "min"]:
        raise ValueError("O tipo deve ser 'max' ou 'min'")
    
    # check if funcao_otimizar is a function
    if not callable(funcao_otimizar):
        raise TypeError("A função de otimização deve ser uma função")

    x_otimo = None
    y_otimo = None

    if tipo == "max":
        valor_otimo = float("-inf")
    else:
        valor_otimo = float("inf")
    
    for i in range(qtd_niveis):
        for x in dominio:
            y = valor_soma - x
            valor = funcao_otimizar(x, y)

            if valor > valor_otimo and tipo == "max" or \
                valor < valor_otimo and tipo == "min":
                valor_otimo = valor
                x_otimo = x
                y_otimo = y
        
        dominio = np.arange(x_otimo - 1, x_otimo + 2, 10 ** -(i + 1))
    
    return x_otimo, y_otimo, valor_otimo
```
