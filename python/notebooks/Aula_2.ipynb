```python
j = 0 

while True:
    if j != 2:
        print(j)
    else:
        print("pulando o numero 2")
        j = j + 1
        continue
    print(j)
    if j == 3:
        print("chegou onde eu queria, dando break")
        break
    j = j + 1
print("acabou sa merda")



```

```python
conjunto = [0,1,2,3,4,5]

for num in conjunto: # o num e uma variavel intermediaria, sera o nome generico dado a cada elemento da lista
    print(f"o valor de x e: {num}")
```

```python
for num in range(0, 10, 1):
    print(f"o numero vale {num}")
```

```python
for num in range(0, 10, 1):
    if num == 2:
        continue #vai pular o 2
    print(f"o numero vale {num}")
```

```python
for num in range(0, 10, 1):
    if num == 4:
        break #vai pular fora no 4
    print(f"o numero vale {num}")
```

```python
nomes = ["almas", "maria", "joao", "pedro"]
indice = 0 
for nome in nomes:  #o nome da variavel intermediaria e uma variavel global, ou seja, ela existe fora do for tambem
    print(f"o indice do nome {nome} e: {indice}")
    indice = indice + 1
```

```python
#de uma forma mais inteligente, usando a funcao enumerate, que ja retorna o indice e o valor do elemento da lista
print(enumerate(nomes)) # forma compactada
print(list(enumerate(nomes)))
```

```python
nomes_enumerados = list(enumerate(nomes))
print(nomes_enumerados)
i = 0 

for i,nome in nomes_enumerados:
    print(i,nome)
```
