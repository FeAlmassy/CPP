```python
class Vetor2D:
     # aqui pode entrar as propriedades da classe, que sao aquelas propriedades fixas
     #tipo o pi, caso a classe fosse sobre circunferencias, CONSTANTES

     #aqui entrar as propriedades genericas (ai entra o self, um generalizador), que vao mudar, no caso da circunferencia, pode ser o raio
     # onde o circunferencia ta centrada etc, no nosso caso as coordenadas do vetor
    def __init__(self, i, j):
        self.i = i
        self.j = j

    #aqui entram os metodos da classe, ou seja, todas as acoes que a classe pode fazer, ela usa o self
    #pra puxar as infos genericas 
    def norma(self):
        return (self.i**2 + self.j**2) ** (1/2)
    
    def prod_escalar(self, vetor_2):
        return self.i * vetor_2.i + self.j * vetor_2.j
```

```python
vetorum = Vetor2D(1,2)
vetordois = Vetor2D(2,5)
```

```python
print(vetorum.norma())
print(vetordois.norma())
```

```python
print(vetordois.prod_escalar(vetorum))
```

```python
class Circunferencia:
    # Propriedade de classe, constante, ou seja, nao precisa do self
    pi = 3.141592

    # Propriedades genéricas, variaveis, ou seja, precisa do self
    def __init__(self, raio):
        self.raio = raio

    # ======================================================================
    
    # Métodos, ou seja, acoes que a classe de vetores pode fazer 
    def area(self):
        return Circunferencia.pi * self.raio ** 2
    
    # Métodos, ou seja, acoes que a classe de vetores pode fazer 
    def perimetro(self):
        return 2 * Circunferencia.pi * self.raio
    
    # Esse é o método que faz o print,
    def __repr__(self):
        return f"Circunferência(raio={self.raio})"
```

```python
c1 = Circunferencia(2)

print(c1.area())

print(c1)
```
