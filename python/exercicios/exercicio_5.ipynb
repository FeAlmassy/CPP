```python
nota = int(input("digite a nota do aluno"))

if nota < 0:
    print("digite uma nota maior que zero")
elif 0 <= nota <= 10:
    if nota == 0:
        resultado = "F"
    elif 0 < nota <= 4:
        resultado = "D"
    elif 4 < nota <= 6:
        resultado = "C"
    elif 6 < nota <= 7:
        resultado = "B"
    elif 7 < nota <= 8:
        resultado = "A"
    elif 8 < nota <= 9:
        resultado = "A+"
    else:
        resultado = "S"
    print(f"A nota da prova foi {resultado}")
else:
    print("digite uma nota menor que dez, jegue")
```
