```python
def calculadora(num_1, num_2) -> dict:
    if not isinstance(num_1, (float, int)) or not isinstance(num_2, (float, int)):
        raise TypeError(f"Você digitou {num_1} e {num_2} digite somente números.")

    if num_2 == 0:
        raise ZeroDivisionError("num_2 não pode ser zero pois é usado como denominador.")

    soma = num_1 + num_2
    subtracao = num_1 - num_2
    multiplicacao = num_1 * num_2
    divisao = num_1 / num_2

    print(f"Soma: {soma} | Subtração: {subtracao} | Multiplicação: {multiplicacao} | Divisão: {divisao}")

    return {"soma": soma, "subtracao": subtracao, "multiplicacao": multiplicacao, "divisao": divisao}
```

```python
calculadora(4,4)
```
