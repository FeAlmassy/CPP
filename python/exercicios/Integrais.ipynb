# Evoluindo um código
 Objetivo, fazer uma **calculadora de integrais** via método de soma de Rienmann
 Porém cada célula o código vai **ficando cada vez mais sofisticado**, começando bem cru e indo até um nível bem legal

##### Código V1 - Usando as ferramentas mais iniciais do python
###### Aqui foi usado a definicao de função mais generalista e o for, percorrendo de maneira automatica 

```python
a = -2 #input
b = 2 #input
tamanho_eixo_x = b - a
qnt_ret = 10000 #input
dx = tamanho_eixo_x/qnt_ret
lista_areas = []

def funcao(x:float):
    return x ** 2

for ret in range(qnt_ret):
    area = dx * funcao(a + ret*dx)
    lista_areas.append(area)
print(sum(lista_areas))
```

##### Código V2 - Usando uma nova notação de função
###### Aqui foi usado a definicao de função mais curta e de certa forma, matenática ... ainda com o for

```python
a = -2 #input
b = 2 #input
tamanho_eixo_x = b - a
qnt_ret = 10000 #input
dx = tamanho_eixo_x/qnt_ret
lista_areas = []

f = lambda x: x ** 2 #input

for ret in range(qnt_ret):
    area = dx * f(a + ret*dx)
    lista_areas.append(area)
print(sum(lista_areas))
```

##### Código V3 - Trocando o for pelo while
###### Trocamos o for pelo while pra treinar a troca de iteração de forma manual, coisa que o for faz sozinho e o while precisa de uma variável de iteração

```python
a = -2 #input
b = 2 #input
tamanho_eixo_x = b - a
qnt_ret = 10000 #input
dx = tamanho_eixo_x/qnt_ret
lista_areas = []

f = lambda x: x ** 2 #input

i = 0
while i < qnt_ret: 
    area = dx * f(a + i*dx)
    lista_areas.append(area)
    i = i + 1
print(sum(lista_areas))
```

##### Código V4 - Usando o try e o except pra mapear erros
###### Aqui implementamos o try pra tentar fazer a coleta dos inputs, caso ele consigo o codigo segue e a conta e feita. Porém caso de erro na coleta dos inputs, ja temos o erro por divisao por zero mapeado, e o erro de digitar algo que nao seja um float, caso seja algum outro erro ele devolve uma mensagem de erro generica. Fazemos isso pra evitar que na tela do cliente, caso de erro, apareça a mensagem do debugger no python, assim fica mais profissional

```python
try:
    a = -2 #input
    b = 2 #input
    tamanho_eixo_x = b - a
    qnt_ret = 10000 #input
    dx = tamanho_eixo_x/qnt_ret
    lista_areas = []

except ZeroDivisionError:
    print("A quantidade de retangulos tem que ser maior que zero ")
except ValueError:
    print("digite um numero")
except Exception as e:
    print("deu algum erro q n to ligado erro 404")


f = lambda x: x ** 2 #input

i = 0
while i <= qnt_ret: 
    area = dx * f(a + i*dx)
    lista_areas.append(area)
    i = i + 1
print(sum(lista_areas))

```

##### Código V5 - Implementando o Raise e o exit()
###### Aqui implementamos o raise que, diferente do try que fica esperando os erros acontecerem ja vai la e mostra uma mensagem caso o erro aconteça. E a função exit que faz com que o programa se encerre caso ele encontre algum erro

```python
#INPUTS
try:
    a = float(input("Início (a): "))
    b = float(input("Fim (b): "))
    qnt_ret = int(input("Quantidade de retângulos: "))
    
    if qnt_ret <= 0:
        raise ZeroDivisionError("A quantidade de ret deve ser maior que zero")

except ValueError:
    print("Erro: Por favor, digite apenas números.")
    exit() # Interrompe aqui para não tentar calcular o que não existe
except ZeroDivisionError:
    print("Erro: A quantidade de retângulos deve ser maior que zero.")
    exit()

#MODELAGEM
tamanho_eixo_x = b - a
dx = tamanho_eixo_x / qnt_ret
lista_areas = []
f = lambda x: x**2

i = 0
while i < qnt_ret:
    x_atual = a + i * dx
    area = dx * f(x_atual)
    lista_areas.append(area)
    i += 1

print(f"Resultado da integral: {sum(lista_areas)}")
```
