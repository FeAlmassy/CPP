```python
#verificar se tem limite

def limite(f:Callable, ponto:float, delta:float = 0.001, tolerancia:float = 0.15) -> float:
    """
    Função criada pra verificar se existe limite da funcao em um certo ponto e se existir, calcular
    Recebe uma funcao, um ponto, e um delta, (variacao pra cima e pra baixo do limite)

    pro calculo do limite, se ele existir, calculamos a media dos limites laterais        
    """

    if not isinstance(ponto, (int, float)) or not isinstance(delta, (int, float)) or not isinstance(tolerancia, (int, float)):
        raise TypeError("digite float tanto pro ponto quanto pro delta")
        
    limite_esq = f(ponto - delta)
    limite_dir = f(ponto + delta)

    if abs(limite_esq - limite_dir) <= tolerancia:
        limite_final = (limite_esq + limite_dir)/2
        return(f"O limite existe e é: {limite_final}")
        
    else:
        return(f"O limite não existe pois os limites laterais não são iguais, {limite_esq} =! {limite_dir}")
```
