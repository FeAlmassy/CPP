```python
numeros = [1, 2, 3, 4, 5, 6]
num_pares = []

for num in numeros:
    if not isinstance(num, int):
        raise TypeError("a lista deve ter somente valores inteiros")
    
    if num < 0:
        raise ValueError("nao podem ter numeros negativos")

    if num % 2 == 0:
        num_pares.append(num)

print(num_pares)
print(sum(num_pares))
```

```python
numeros = [1, 2, 3, 4, 5, 6]
num_pares = []

try:
    if not isinstance(numeros, list):
        raise TypeError("A entrada deve ser uma lista.")

    for num in numeros:
        if not isinstance(num, (int, float)):
            raise TypeError(f"O valor {num} nao é numerico.")
        
        if num < 0:
            raise ValueError("Nao podem existir numeros negativos.")

        if num % 2 == 0:
            num_pares.append(num)

    print("Numeros pares:", num_pares)
    print("Soma:", sum(num_pares))

except TypeError as e:
    print("Erro de tipo:", e)

except ValueError as e:
    print("Erro de valor:", e)

except Exception as e:
    print("Erro inesperado:", e)
```
