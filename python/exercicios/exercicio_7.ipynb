```python
tentativas_max = 3
senha_correta = "dawdawd"
tentativas = ["dog", "pig", "boi"]

status = {
    "sucesso": False,
    "tentativa_numero": 0,
    "bloqueado": False
}

qnt_tentativas = 0

for senha in tentativas:


    if senha == senha_correta:
        status["sucesso"] = True
        status["tentativa_numero"] = qnt_tentativas
        status["bloqueado"] = False
        break

    if senha != senha_correta:
        status["sucesso"] = True
        status["tentativa_numero"] = qnt_tentativas
        status["bloqueado"] = False

        break

        

    if qnt_tentativas >= tentativas_max:
        print(f"vc digitou a senha incorreta, restam: {qnt_tentativas} tentativas restantes")
        qnt_tentativas = qnt_tentativas + 1
        break
    

```
