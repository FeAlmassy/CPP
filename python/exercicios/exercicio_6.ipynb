```python
lista = [1,4,9,3,5,8,12,8,22,-2]
maior_valor = lista[0] #por hipotese
menor_valor = lista[0] #por hipotese

for num in lista:
    if num > maior_valor:
        maior_valor = num
    elif num < menor_valor:
        menor_valor = num
    else:
        continue

print(f"O maior valor encontrado na lista vale: {maior_valor}")
print(f"O menor valor encontrado na lista vale: {menor_valor}")
```

```python
lista = [1,4,9,3,5,8,12,8,22,-2]
maior_valor = lista[0]
menor_valor = lista[0]

i = 0


while i < len(lista):

    if not isinstance(lista[i], (float,int)):
        raise ValueError("digite somente numeros jegue")

    if lista[i] > maior_valor:
        maior_valor = lista[i]

    if lista[i] < menor_valor:
        menor_valor = lista[i]

    i = i + 1

print(f"O maior valor encontrado na lista vale: {maior_valor}")
print(f"O menor valor encontrado na lista vale: {menor_valor}")
```
