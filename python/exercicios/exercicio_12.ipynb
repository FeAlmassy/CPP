##### Código V1 - Maneira gorila mas funciona
###### única novidade é o usa da função sorted pra deixar uma lista em ordem do menor pro maior

```python
lados_triangulo = sorted([5,3,3]) #pega do menor pro maior
print(f"Os lados do triangulo sao: {lados_triangulo}")

a,b,c = lados_triangulo
tipo_triangulo = "None"

if a + b > c:
    if a == b == c:
        tipo_triangulo = "Equilatero"
    elif a != b != c:
        tipo_triangulo = "Escaleno"
    else:
        tipo_triangulo = "Isosceles"
    print(f"O triangulo existe, categoria: {tipo_triangulo}")
else:
    print("nao e um triangulo")
```

##### Código V2 - Maneira profissional
###### Fazemos a tratativa das condicoes de contorno usando o raise (erro) pra ja achar ali algum possivel problema. O resto da modelagem segue o msm

```python
lados_entrada = [3,4,5]

#Verificacao dos dados
for lado in lados_entrada:
    if not isinstance(lado, (int, float)):
        raise TypeError(f"O valor '{lado}' nao e um numero valido!")
    if lado <= 0:
        raise ValueError(f"O lado {lado} e invalido! Lados devem ser maiores que zero.")

#Modelagem
lados = sorted(lados_entrada)
a, b, c = lados

if a + b > c:
    if a == b == c:
        tipo_triangulo = "Equilatero"
    elif a != b != c:
        tipo_triangulo = "Escaleno"
    else:
        tipo_triangulo = "Isosceles"
    print(f"O triangulo existe, categoria: {tipo_triangulo}")
else:
    print("nao e um triangulo")
```
