```python
def integral(funcao, a:float, b:float, qnt_ret:int = 1000):
    
    if not callable(funcao):
        raise TypeError("digite uma funcao seu jegue")
    if not isinstance(a, (float, int)) or not isinstance(b, (float, int)):
        raise TypeError("digite numero jegue")
    if qnt_ret <= 0:
        raise ValueError("o qnt de retangulos deve ser maior que zero")
    
    largura = (b - a) / qnt_ret
    return sum(funcao(a + i * largura) * largura for i in range(qnt_ret))
    
```

```python
integral(quadratica, 0 ,2, 100)
```

```python
import math

def dist(p1:list, p2:list):
    """
    obj: calcular a distancia entre dois pontos via geometria analitica

    entrada 1: Par de coordenadas em formato de lista 
    entrada 2: Par de coordenadas em formato de lista 

    saida: a distancia entre os dois pontos
    """
    c1 = p2[1] - p1[1]
    c2 = p2[0] - p1[0]
    d = math.sqrt(c1 ** 2 + c2 ** 2)
    return d


print(dist([2,2], [3,3]) )
#ou
print(dist(p1=[2,2], p2=[3,3]))
```
