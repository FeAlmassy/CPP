```python
texto = "python"

vogais = "aeiouAEIOU"
consoantes = "bcdfghjklmnpqrstvwxyzBCDFGHJKLMNPQRSTVWXYZ"

qtd_vogais = []
qtd_consoantes = []
qtd_tralha = []

for letra in texto:
    if letra in vogais:
        qtd_vogais.append(letra)
    elif letra in consoantes:
        qtd_consoantes.append(letra)
    else:
        qtd_tralha.append(letra)

print(qtd_vogais)
print(qtd_consoantes)
print(qtd_tralha)
```
